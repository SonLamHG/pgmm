# kernel_sources probe

Does one kernel's output mount inside another? If so the prepared dataset
never has to leave Kaggle, and D16's local round-trip is unnecessary for it.

Motivation: `kernels output` fetches files one at a time. With ~228k prepared
frames it stalled at 1 file after 6 minutes. Cost was budgeted in gigabytes
(fine) but the real currency is file count (not fine).

```python
from pathlib import Path
INPUT = Path('/kaggle/input')
print('mounted:', [p.name for p in INPUT.iterdir()] if INPUT.exists() else 'NOTHING')
jpgs = list(INPUT.rglob('frame_*.jpg'))
print('frames visible:', len(jpgs))
if jpgs:
    print('sample:', jpgs[0].relative_to(INPUT))
    print('KERNEL_SOURCES WORKS -- no transfer needed')
else:
    print('KERNEL_SOURCES DID NOT MOUNT -- fall back to a single-archive round-trip')
idx = list(INPUT.rglob('index.json'))
print('index.json:', idx[0].relative_to(INPUT) if idx else 'absent')
```
