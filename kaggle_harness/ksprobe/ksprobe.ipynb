# What did the preprocess kernel's output actually keep?

The kernel logged `output will contain: ['__notebook__.ipynb', 'lsa64_prepared']`
with 264,831 frames on disk. Yet `kernels output` returns a 0-byte zip and a
kernel_sources mount shows no frames. So the question is what Kaggle *saved*,
not what the kernel *wrote* -- a file-count limit is the suspect.

```python
from pathlib import Path
INPUT = Path('/kaggle/input')
print('top level:', [p.name for p in INPUT.iterdir()] if INPUT.exists() else 'NOTHING')
print()
n = 0
for p in sorted(INPUT.rglob('*')):
    n += 1
    if n <= 40:
        kind = 'd' if p.is_dir() else f'{p.stat().st_size}B'
        print(f'  [{kind}] {p.relative_to(INPUT)}')
print()
print('total entries under /kaggle/input:', n)
print('jpg count:', sum(1 for _ in INPUT.rglob('*.jpg')))
```
